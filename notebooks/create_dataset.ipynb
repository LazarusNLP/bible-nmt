```python
from datasets import load_dataset
import os
output_dir = "/data/projects/punim0478/setiawand/dali/dataset"

def process_tetun_wili_2018():
    with open(os.path.join(output_dir, "tetum_wili_2018.txt"), "w") as f:
        for datum in tetun_wili_2018["train"]:
            if datum["label"] == 139:
                f.write(datum["sentence"] + "\n")
                
def process_tetun_lti_langid():
    with open(os.path.join(output_dir, "tetum_lti_langid.txt"), "w") as f:
        for split in tetun_lti_langid.keys():
            for datum in tetun_lti_langid[split]:
                if datum["language"] == "tet":
                    f.write(datum["text"] + "\n")
                    
def process_tetum_wiki():
    with open(os.path.join(output_dir, "tetum_wiki.txt"), "w") as f:
        for datum in tetun_wiki:
            f.write(datum["text"] + "\n")

def process_tetum_glot_story():
    with open(os.path.join(output_dir, "tetum_glot_story.txt"), "w") as f:
        for split in tetun_glot_story.keys():
            for datum in tetun_glot_story[split]:
                if "tet" in datum["language"].lower():
                    f.write(datum["text"] + "\n")
            
            
tetun_wiki = load_dataset(
        "sabilmakbar/indo_wiki", "indowiki_dedup_all", split="tet", trust_remote_code=True
    )
tetun_bible = "/data/projects/punim0478/setiawand/ebible/corpus/tet-tet.txt"
tetun_bloom_lm = load_dataset("/data/projects/punim0478/setiawand/seacrowd-datahub/seacrowd/sea_datasets/bloom_lm", "bloom_lm_tet_source", trust_remote_code=True)
tetun_bloom_vi= load_dataset("/data/projects/punim0478/setiawand/seacrowd-datahub/seacrowd/sea_datasets/bloom_vist", "bloom_vist_tet_source", trust_remote_code=True)
tetun_bloom_caption = load_dataset("/data/projects/punim0478/setiawand/seacrowd-datahub/seacrowd/sea_datasets/bloom_captioning", "bloom_captioning_tet_source", trust_remote_code=True)
tetun_glot_story = load_dataset("/data/projects/punim0478/setiawand/seacrowd-datahub/seacrowd/sea_datasets/glotstorybook", trust_remote_code=True) # tetun dili
tetun_lti_langid = load_dataset("/data/projects/punim0478/setiawand/seacrowd-datahub/seacrowd/sea_datasets/lti_langid_corpus", trust_remote_code=True) # tetun dili
tetun_wili_2018 = load_dataset("/data/projects/punim0478/setiawand/seacrowd-datahub/seacrowd/sea_datasets/wili_2018", trust_remote_code=True) # label 139
# tetun_sea_madlad = load_dataset("/data/projects/punim0478/setiawand/seacrowd-datahub/seacrowd/sea_datasets/sea_madlad", "sea_madlad_tet_source", trust_remote_code=True)
```
