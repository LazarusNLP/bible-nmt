```python
# Create comprehensive baseline comparison table
import pandas as pd

# Extract evaluation metrics from each baseline
def extract_metrics(data, baseline_name, metric_type=None):
    if 'eval_metrics' in data:
        metrics = data['eval_metrics']
    elif metric_type and metric_type in data:
        # For LLM baselines with specific metric types
        metrics = data[metric_type]
    else:
        # For LLM baselines, metrics might be at root level
        metrics = data
    
    return {
        'Model': baseline_name,
        'spBLEU': metrics.get('spbleu', 0),
        'chrF++': metrics.get('chrf++', 0)
    }

# Create list of all baselines with proper naming
baselines_data = []

# NLLB-200-600M baselines
baselines_data.append(extract_metrics(web_metric, 'nllb-200-distilled-600M_{NT}'))
baselines_data.append(extract_metrics(web_plus_metric, 'nllb-200-distilled-600M_{NT+Parallel}'))

# LLM baselines - use the correct metric sections
baselines_data.append(extract_metrics(llm_translation, 'LLM_translation', 'original_mt_metrics'))
baselines_data.append(extract_metrics(llm_post_edit, 'LLM_post_editing', 'post_edited_metrics'))

# Create DataFrame
baseline_comparison_df = pd.DataFrame(baselines_data)

# Round to 2 decimal places for better readability
baseline_comparison_df = baseline_comparison_df.round(2)

baseline_comparison_df
```
