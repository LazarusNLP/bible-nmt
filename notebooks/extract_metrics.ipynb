```python
import json
iso_codes = ["tet", "ptu" ,"wrs" , "row" , "nfa" , "aaz" , "lex"]
# iso_codes = ["lex", "wrs", "ptu"]
lang = {
    "ptu": "Bambam",
    "tet": "Tetun",
    "wrs": "Waris",
    "row": "Rote Dela",
    "nfa": "Dhao",
    "aaz": "Amarasi",
    "lex": "Luang",
}

base_dir = "nllb-200-distilled-600M-scripture_files-ind"
results = {"Language": [], "ISO 639-3": [], "Eval Set": [], "BLEU": [], "spBLEU": [], "SacreBLEU": [], "chrF3": [], "chrF++": []}
def get_metrics(iso_code):
    dir_path = f"../{base_dir}-{iso_code}-v2-5000"
    val_results_path = dir_path + "/validation-results.json"
    test_results_path = dir_path + "/test-results.json"
    
    with open(val_results_path, "r") as f:
        data = json.load(f)
        val_results = data["eval_metrics"]
        results["Language"].append(lang[iso_code])
        results["ISO 639-3"].append(iso_code)
        results["Eval Set"].append("validation")
        results["BLEU"].append(val_results["bleu"])
        results["spBLEU"].append(val_results["spbleu"])
        results["SacreBLEU"].append(val_results["sacrebleu"])
        results["chrF3"].append(val_results["chrf3"])
        results["chrF++"].append(val_results["chrf++"])
        
    with open(test_results_path, "r") as f:
        data = json.load(f)
        test_results = data["eval_metrics"]
        results["Language"].append(lang[iso_code])
        results["ISO 639-3"].append(iso_code)
        results["Eval Set"].append("test")
        results["BLEU"].append(test_results["bleu"])
        results["spBLEU"].append(test_results["spbleu"])
        results["SacreBLEU"].append(test_results["sacrebleu"])
        results["chrF3"].append(test_results["chrf3"])
        results["chrF++"].append(test_results["chrf++"])
    

for iso_code in iso_codes:
    get_metrics(iso_code)
```

```python
import pandas as pd
df = pd.DataFrame(results)
df
```
