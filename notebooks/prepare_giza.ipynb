Create parallel sentence for fast_align word-alignment.

```python
import pandas as pd
bible = pd.read_csv("../ebible-corpus/dhao-eng/aligned-eng-dhao-all.csv")
```

```python
import nltk
from nltk.tokenize import word_tokenize

# Download required data (first time only)
nltk.download('punkt')

def tokenize_text(text):
    tokens = word_tokenize(text)
    return ' '.join(tokens)

# Process all data and write to a single file
with open("../ebible-corpus/dhao-eng/train.en", "w") as source_f, open("../ebible-corpus/dhao-eng/train.nfa", "w") as target_f:
    # Process New Testament data
    for _, row in  bible.iterrows():
        source = row['source_text']
        target = row['target_text']
        source_f.write(tokenize_text(source) +  "\n")
        target_f.write(tokenize_text(target) + "\n")
```
