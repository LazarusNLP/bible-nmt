```python
# SOLUTION: Use configs instead of complex split names
def push_language_pair_datasets(source_lang, target_langs, repo_name):
    """Process and push each language pair as a separate config to the same repo"""
    
    for i, tgt_lang in enumerate(target_langs):
        print(f"\n[{i+1}/{len(target_langs)}] Processing {source_lang} -> {tgt_lang}...")
        
        try:
            # Load and process the dataset for this language pair
            dataset = load_ebible_corpus(source_lang, tgt_lang)
            
            # Create config name (use underscore, not hyphen!)
            config_name = f"{source_lang}_{tgt_lang}"
            
            print(f"  Dataset structure:")
            for split_name, split_data in dataset.items():
                print(f"    {split_name}: {len(split_data):,} examples")
            
            # Push to hub with config name - this is the key!
            dataset.push_to_hub(
                repo_id=repo_name,
                config_name=config_name,  # Each language pair = separate config
                private=False
            )
            
            print(f"  ✅ Successfully pushed {config_name}")
            
        except Exception as e:
            print(f"  ❌ Error processing {tgt_lang}: {e}")
            continue
    
    print(f"\n🎉 All done! Dataset available at: https://huggingface.co/datasets/{repo_name}")

# Set your repository name here
REPO_NAME = "your-username/bible-nmt-multilingual"  # Change this!

print("Ready to push! Set REPO_NAME and uncomment the line below:")
print("# push_language_pair_datasets(source_lang, target_lang, REPO_NAME)")
```
