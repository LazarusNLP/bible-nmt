Create parallel sentence for fast_align word-alignment.

```python
import pandas as pd
nt = pd.read_csv("../ebible-corpus/luang/aligned-eng-luang-nt.csv")
ot = pd.read_csv("../ebible-corpus/luang/aligned-eng-luang-ot.csv")
stories = pd.read_csv("../ebible-corpus/luang/luang-english-parallel.csv")
```

```python
import nltk
from nltk.tokenize import word_tokenize

# Download required data (first time only)
nltk.download('punkt')

def tokenize_text(text):
    tokens = word_tokenize(text)
    return ' '.join(tokens)

# Process all data and write to a single file
with open("../ebible-corpus/luang/aligned-eng-luang-all.txt", "w") as f:
    # Process New Testament data
    for _, row in nt.iterrows():
        source = row['source_text']
        target = row['target_text']
        f.write(tokenize_text(source) + " ||| " + tokenize_text(target) + "\n")
    
    # Process Old Testament data
    for _, row in ot.iterrows():
        source = row['source_text']
        target = row['target_text']
        f.write(tokenize_text(source) + " ||| " + tokenize_text(target) + "\n")
    
    # Process stories data
    for _, row in stories.iterrows():
        source = row['source_text']
        target = row['target_text']
        f.write(tokenize_text(source) + " ||| " + tokenize_text(target) + "\n")
```
