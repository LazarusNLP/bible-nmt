Create parallel sentence for fast_align word-alignment.

```python
with open("../ebible-corpus/dhao/test.src.detok.txt") as f:
    src = f.read().splitlines()

with open("../ebible-corpus/dhao/test.trg.detok.txt") as f:
    tgt = f.read().splitlines()
```

```python
import nltk
from nltk.tokenize import word_tokenize

# Download required data (first time only)
nltk.download('punkt')

def tokenize_text(text):
    tokens = word_tokenize(text)
    return ' '.join(tokens)


with open("../ebible-corpus/dhao/aligned-ind-nfa-ot.txt", "w") as f:
    for source, target in zip(src, tgt):
        f.write(tokenize_text(source) + " ||| " + tokenize_text(target) + "\n")
```
